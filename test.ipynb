```python
def show(name,img):
    cv2.imshow(name,img)

    cv2.waitKey(5000)
    cv2.destroyAllWindows()
```

```python
show("image1",img)
```

```python
img = cv2.imread('test.jpg',cv2.IMREAD_GRAYSCALE)
show("image2",img)
```

```python
vc = cv2.VideoCapture('test.mp4')
```

```python
# 检查是否成功打开
if vc.isOpened():
    open,frame = vc.read() #如果成功打开，返回open(一个布尔值)和frame(第一帧的图像矩阵图)
else:
    open = False
```

```python
while open:
    ret,frame = vc.read()
    if frame is None: #如果播放完了
        break
    if ret == True:
        gray = cv2.cvtColor(frame,cv2.COLOR_BGR2GRAY)
        cv2.imshow('result',gray)
        if cv2.waitKey(10) & 0xFF == 27:
            break

vc.release
cv2.destroyAllWindows()
```
