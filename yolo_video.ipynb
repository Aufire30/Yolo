```python
video = cv2.VideoCapture("test.mp4")
num = 0             # 计数器
save_step = 30      # 间隔帧
while True:
    ret,frame = video.read()
    if not ret:
        break
    num += 1
    if num % save_step == 0:
        cv2.imwrite("video_img/" + str(num//30) + ".jpg",frame)
```
